## Configure Apertures, Telescope Coordinate System

To simplify this Notebook, the `defineApertures` function from `selectSIAF` is essentially a lookup table, translating user-friendly instrument names into the correct corresponding apertures. 

This same function also generates the correct `V2Ref` and `V3Ref` offset angles. These V2/V3 values answer the question "what angles are there between this instrument and the line-of-sight of the telescope?". This is quite useful since, barring a catastrophic event, the instruments do not change their alignment relative to the line-of-sight.


```python
# Set up aperture list and reference coordinates
apertureList, V2Ref, V3Ref = defineApertures(selectedTelescope, selectedInstrument, selectedAperture)
```

## Telescope Position Angle

The last variable we need to set before using `pysiaf` is the telescope position angle. From the docstring of the `attitude_matrix` function:
>Position angle of V3 axis at nu2,nu3 measured from North to East (default unit is degree)

In plain English, this is the telescope roll angle; with the telescope fixed on the target, it rotates around its V1 (line-of-sight) axis. Modifying this value will cause the footprints to "roll" around the target.

```python
# can be any value between 0 and 360
telescopePositionAngle = 91.5 #91.5 #alternative=92
```

```python
# Second WFI Orientation!
telescopePositionAngle2 = 271.5 #271.5 #alternative=272
```

## Generate The Attitude Matrix

At this point, we've generated footprints in the telescope's field of view. Now, we need to translate this into a view on the sky, in RA/Dec coordinates. This is handled by the `attitude_matrix` function, which is doing some matrix multiplication behind the scenes.

```python
# Set telescope pointing matrix: "attitude matrix"
attmat = pysiaf.utils.rotations.attitude_matrix(V2Ref, V3Ref, targetRa, targetDec, telescopePositionAngle)
```

```python
# Set telescope pointing matrix: "attitude matrix"
attmat2 = pysiaf.utils.rotations.attitude_matrix(V2Ref, V3Ref, targetRa2, targetDec2, telescopePositionAngle2)
```

## Calculate Sky Regions

The final step is to convert to the [IVOA Space-Time Coordinate Metadata Linear String Implementation (STS-C)](https://ivoa.net/documents/STC-S/). Do not fear this verbose name; it's a method of standardizing footprints in string formats. The specifics depend on the particular shape used. For the curious, the `computeStcsFootprint` function from `selectSIAF` is what calcuates these strings in this Notebook.

Since `Aladin` knows how to parse this format, it will be able to plot it on a sky background of our choice. It is not particularly human readable, but we'll print it out anyway.

```python
# Loop through aperture list
# Transform to sky coordinates, build the total footprints
combinedSregion1 = ''
for apertureSiaf in apertureList:
    apertureSiaf.set_attitude_matrix(attmat)
    xVertices, yVertices = getVertices(apertureSiaf)
    
    if (xVertices is not None and yVertices is not None):
        skyRa, skyDec = apertureSiaf.idl_to_sky(xVertices, yVertices)
        apertureSregion1 = computeStcsFootprint(apertureSiaf, skyRa, skyDec)
        combinedSregion1 += apertureSregion1
        
#print('\n'+combinedSregion1)
```

```python
from shapely.geometry import Polygon

def chips_to_polygons(chip_coords):
    return [Polygon(coords) for coords in chip_coords]

def overlap_area_fraction_union(chips1, chips2):
    polys1 = chips_to_polygons(chips1)
    polys2 = chips_to_polygons(chips2)

    # total areas of the two pointings
    area1 = sum(p.area for p in polys1)
    area2 = sum(p.area for p in polys2)

    # compute total overlap area
    overlap_area = 0.0
    for p1 in polys1:
        for p2 in polys2:
            inter = p1.intersection(p2)
            if not inter.is_empty:
                overlap_area += inter.area

    # union area = A1 + A2 - overlap
    union_area = area1 + area2 - overlap_area

    frac = overlap_area / union_area if union_area > 0 else 0.0
    return frac, overlap_area, union_area, area1, area2
```

```python
frac, overlap_area, union_area, area1, area2 = overlap_area_fraction_union(chips1, chips2)

print(f"Overlap area  = {overlap_area:.4f} deg^2")
print(f"Total area    = {union_area:.4f} deg^2")
print(f"Fractional overlap (relative to total) = {frac:.2%}")
```

```python
import astropy.units as u

def transform_to_galactic(chips):
    """Transform chip corners from (RA,Dec) to Galactic (l,b) in degrees."""
    gal_chips = []
    for chip in chips:
        ra_vals, dec_vals = zip(*chip)
        coords = SkyCoord(ra=ra_vals*u.deg, dec=dec_vals*u.deg, frame='icrs')
        l = coords.galactic.l.deg
        b = coords.galactic.b.deg
        # Wrap longitudes to [-180,180)
        l = ((l + 180) % 360) - 180
        gal_chips.append(list(zip(l, b)))
    return gal_chips

def plot_pointings_cartopy(chips1, chips2, center_l=0, center_b=0, width=10, height=10):
    """
    Plot pointings using Cartopy with zoom on galactic center region.
    
    Parameters:
    - chips1, chips2: List of chip polygons
    - center_l, center_b: Center coordinates in degrees
    - width, height: View area dimensions in degrees
    """
    # Transform coordinates
    chips1_gal = transform_to_galactic(chips1)
    chips2_gal = transform_to_galactic(chips2)
    
    # Create polygons
    polys1 = [Polygon(coords).buffer(0) for coords in chips1_gal]
    polys2 = [Polygon(coords).buffer(0) for coords in chips2_gal]

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection=ccrs.Mollweide(central_longitude=center_l))
    
    # Set the extent to zoom in
    ax.set_extent([center_l - width/2, center_l + width/2,
                   center_b - height/2, center_b + height/2],
                  crs=ccrs.PlateCarree())
    
    # Custom gridlines with labels at bottom
    gl = ax.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                      linewidth=0.5, color='gray', alpha=0.5)
    
    # Configure where labels appear
    gl.top_labels = False    # Turn off top labels
    gl.right_labels = False  # Turn off right labels
    gl.bottom_labels = True  # Keep bottom labels
    gl.left_labels = True    # Keep left labels
    
    # Optional: Control label spacing
    #gl.xlocator = plt.FixedLocator(np.arange(-180, 181, 1))  # Longitude ticks every 5°
    #gl.ylocator = plt.FixedLocator(np.arange(-90, 91, 1))    # Latitude ticks every 5°

    # For fixed ticks regardless of zoom level
    gl.xlocator = plt.FixedLocator([-2, 0, 2])  # Specific longitudes
    gl.ylocator = plt.FixedLocator([-5, -2, 0, 2, 5])  # Specific latitudes
    
    # Plot Region 1 (blue)
    for p in polys1:
        if not p.is_empty:
            l, b = zip(*p.exterior.coords)
            ax.fill(l, b, transform=ccrs.PlateCarree(),
                    alpha=0.4, fc='blue', ec='black', lw=0.5, label="Spring")
    
    # Plot Region 2 (red)
    for p in polys2:
        if not p.is_empty:
            l, b = zip(*p.exterior.coords)
            ax.fill(l, b, transform=ccrs.PlateCarree(),
                    alpha=0.4, fc='red', ec='black', lw=0.5, label="Fall")
    
    # Plot Overlap (purple)
    for p1 in polys1:
        for p2 in polys2:
            inter = p1.intersection(p2).buffer(0)
            if not inter.is_empty:
                if inter.geom_type == 'Polygon':
                    l, b = zip(*inter.exterior.coords)
                    ax.fill(l, b, transform=ccrs.PlateCarree(),
                            alpha=0.7, fc='purple', ec='none', label="Overlap")
                elif inter.geom_type == 'MultiPolygon':
                    for sub in inter.geoms:
                        l, b = zip(*sub.exterior.coords)
                        ax.fill(l, b, transform=ccrs.PlateCarree(),
                                alpha=0.7, fc='purple', ec='none', label="Overlap")
    
    # Title and legend
    #ax.set_title(f"Roman WFI Pointings (Galactic Coordinates)\nCenter: (l={center_l}°, b={center_b}°)")
    
    # Remove duplicate legend entries
    handles, labels = ax.get_legend_handles_labels()
    unique = dict(zip(labels, handles))
    ax.legend(unique.values(), unique.keys(), loc="upper right", ncol=3, frameon=False)

    ax.invert_xaxis()

    #ax.plot(359.944, -0.046, marker='o', ms=2, color='yellow', transform=ccrs.PlateCarree()) #GC
    #ax.plot(0.16515613, -0.06217547, marker='o', ms=2, color='yellow', transform=ccrs.PlateCarree()) #Quintuplet
    #ax.plot(0.12261477, 0.01714483, marker='o', ms=2, color='yellow', transform=ccrs.PlateCarree()) #Arches
    
    #ax.plot(0.03361856, -0.25525722, marker='*', ms=2, color='yellow', transform=ccrs.PlateCarree()) #Bright H~3

    #BONUS OBJECT: One published microlensing planet: UKIRT-2017-BLG-001Lb
    #ax.plot(359.8781959, -0.3290687, marker='X', ms=2, color='orange', transform=ccrs.PlateCarree())
    

    #ax.text(0.9,0.7, r"X Offset = -0.15$\degree$", transform=ccrs.PlateCarree())
    #ax.text(0.9,0.65, r"Y Offset = +0.01$\degree$", transform=ccrs.PlateCarree())
    #ax.text(0.9,0.6, r"Frac. Overlap = 0.753", transform=ccrs.PlateCarree())

    #ax.text(0.9,0.5, r"Spring PA = 90$\degree$", transform=ccrs.PlateCarree())
    #ax.text(0.9,0.45, r"Fall PA = 270$\degree$", transform=ccrs.PlateCarree())
    
    plt.tight_layout()

    plt.savefig("GC_overlap_regs.png", dpi=500)
    plt.savefig("GC_overlap_regs.pdf")
    plt.show()
```
